## 3. Success metric

*One metric you can defend. What number means 'good'?*

##I will use **Precision at K** (e.g., Precision@50) as my success metric. For an SEO team, 'good' means that if they audit the top 50 pages identified by the model, a high percentage of those pages actually show signs of being stagnant or low-quality. This focuses the metric on the 'usefulness' of the priority queue rather than just overall model accuracy.

```python
# Showing the concept: we prioritize items with the steepest downward trend
top_risk_items = df.sort_values(by='trend_pct', ascending=True).head(50)
print(f"Top 50 items identified by steepest downward trend:")
print(top_risk_items[['content_id', 'trend_pct']].head())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

#A fixed rule (like 'if impressions drop 20%, flag it') is too blunt and creates noise because it fails to consider the **context** of the content. An ML approach beats this because it can weigh multiple signals simultaneously—such as traffic volatility, CTR stability, and content category—to detect complex patterns of stagnation that a simple if-statement would miss. ML helps us distinguish between a temporary seasonal dip and a genuine, long-term performance decline, which a static threshold would incorrectly group together.

```python
# Show the number of features that a human rule would struggle to balance simultaneously
features = [col for col in df.columns if col not in ['content_id', 'client_id']]
print(f"Number of signals (features) the model can weigh simultaneously: {len(features)}")
print("A fixed rule is 'if-this-then-that', but ML learns the non-linear interaction between these 40+ signals.")
```
