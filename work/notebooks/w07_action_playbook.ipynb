## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

- **Action Queue Logic:** Ranks content assets by combining staleness, traffic decay risk, and commercial value (`search_volume` and `cpc`).
- **Reason Codes:**
  1. `STALE_HIGH_VALUE` — Content is older than 180 days with high search volume and zero recent optimizations.
  2. `AUTHORITY_DECAY` — High-backlink page experiencing impression compression.
  3. `THIN_CONTENT_OPPORTUNITY` — High-impression page with word count below the competitive threshold.

```python
import pandas as pd
import os
from google.colab import userdata
from huggingface_hub import login
from datasets import load_dataset

# Load dataset and prepare the ranked action queue
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
dataset = load_dataset("FlyRank/internship-warehouse", "dim_content")
df = pd.DataFrame(dataset['train'])

# Compute action score
df['action_score'] = (
    df['search_volume'].fillna(0) * 0.4 +
    df['cpc'].fillna(0) * 100 * 0.3 +
    df['backlinks'].fillna(0) * 0.3
)

ranked_df = df.sort_values(by='action_score', ascending=False).reset_index(drop=True)
ranked_df['rank'] = ranked_df.index + 1

print(f"Total actions ranked in queue: {len(ranked_df)}")
display(ranked_df[['rank', 'content_hash_id', 'word_count', 'search_volume', 'action_score']].head(5))
```
